The first witness lives at the last house on "Northwestern Dr"

```sql
SELECT id, name, address_street_name, MAX(address_number) FROM person
WHERE LOWER(address_street_name) = "northwestern dr"
HAVING MAX(address_number) = address_number
```

```sql
SELECT * FROM interview
WHERE person_id IN (
        SELECT id FROM person
        WHERE LOWER(address_street_name) = "northwestern dr"
        ORDER BY address_number DESC
        LIMIT 1
)
```

- Son numéro de membre commence par 48Z
- Il est membre gold
- La plaque de sa voiture comporte les caractères "H42W"
- Il était présent au club le 9 janvier 2018

```sql
SELECT DISTINCT membership_status FROM get_fit_now_member
```

```sql
SELECT id, person_id, name 
FROM get_fit_now_member gfm, get_fit_now_check_in gfc
WHERE gfm.id = gfc.membership_id
AND id LIKE "48Z%"
AND membership_status = "gold"
AND check_in_date = 20180109
```

```sql
SELECT p.id, p.name, plate_number
FROM get_fit_now_member gfm, person p, drivers_license d
WHERE gfm.person_id = p.id
AND p.license_id = d.id
AND gfm.id LIKE "48Z%"
AND membership_status = "gold"
AND plate_number LIKE "%H42W%"
```

```sql
SELECT p.id, name, annual_income, COUNT(*) AS "présence concert"
FROM person p, drivers_license d, income i, facebook_event_checkin f
WHERE p.license_id = d.id
AND p.ssn = i.ssn
AND p.id = f.person_id
AND height BETWEEN 65 AND 67
AND hair_color = "red"
AND gender = "female"
AND car_make = "Tesla"
AND car_model = "Model S"
AND event_name = "SQL Symphony Concert"
AND f.date BETWEEN 20171201 AND 20171231
GROUP BY p.id
ORDER BY annual_income DESC
```
